```python
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

df_cleaned=df.dropna()
print("Cleaned Data:\n")
print(df_cleaned)
```

```python
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

df.fillna(0, inplace=True)
print(df)
```

```python
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

df.fillna(df.mean(), inplace=True)
print(df)
```
